# Multiple imputation by chained equations (MICE)

```python
# Cargar bibliotecas
import seaborn as sns
import pandas as pd
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

```python
# Cargar el dataset de Titanic desde seaborn
titanic = sns.load_dataset('titanic')
```

```python
# Mostrar primeros valores
titanic.head()
```

```python
# Check missing data
print(titanic.isnull().sum())
```

```python
# Vamos a imputar los valores faltantes de la columna 'age' (edad)
# Seleccionamos un subconjunto de columnas para el ejemplo
data = titanic[['age', 'fare', 'pclass', 'sex']]

# Convertimos las variables categóricas en variables dummy (la variable 'sex' es categórica)
data = pd.get_dummies(data, drop_first=True)

data.head(20)
```

```python
# Crear el imputador MICE (IterativeImputer)
imputer = IterativeImputer(max_iter=10, random_state=0)
```

```python
# Imputar los valores faltantes
data_imputed = pd.DataFrame(imputer.fit_transform(data), columns=data.columns)
```

```python
# Mostrar los datos originales con valores faltantes
print("Datos originales (con valores faltantes):")
data.head(30)
```

```python
# Mostrar los datos después de la imputación múltiple (MICE)
print("\nDatos después de la imputación múltiple (MICE):")
data_imputed.head(30)
```

# Exercise 1
Use MICE on the life-expectancy dataset to fill in missing values.
https://www.kaggle.com/datasets/kumarajarshi/life-expectancy-who

```python
# Exercise 1: Load life-expectancy dataset and apply MICE
import pandas as pd
from sklearn.experimental import enable_iterative_imputer  # noqa
from sklearn.impute import IterativeImputer

# Load the life expectancy dataset
life_expectancy = pd.read_csv('Life Expectancy Data.csv')  # Update with correct path if needed

# Display missing data
print("Missing values in Life Expectancy Dataset:")
print(life_expectancy.isnull().sum())

# Separate numeric columns for imputation and keep non-numeric columns
numeric_columns = life_expectancy.select_dtypes(include=['float64', 'int64']).columns
non_numeric_columns = life_expectancy.select_dtypes(exclude=['float64', 'int64'])

# Apply MICE on the numeric columns only
life_expectancy_numeric = life_expectancy[numeric_columns]
mice_imputer = IterativeImputer(random_state=0)
life_expectancy_imputed_numeric = pd.DataFrame(mice_imputer.fit_transform(life_expectancy_numeric), columns=numeric_columns)

# Combine the imputed numeric data with the non-numeric columns
life_expectancy_imputed = pd.concat([non_numeric_columns.reset_index(drop=True), life_expectancy_imputed_numeric], axis=1)

# Verify imputation results
print("Missing values after MICE imputation:")
print(life_expectancy_imputed.isnull().sum())

# Display the first few rows of the imputed data
life_expectancy_imputed.head()
```
